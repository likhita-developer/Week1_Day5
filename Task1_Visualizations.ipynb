```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("clean_data.csv")

print("Dataset loaded successfully!")
print(df.head())
```

```python
# 1. BAR CHART - Employee Count by Department

import os

# Create figures folder
os.makedirs("reports/figures", exist_ok=True)

department_count = df["department"].value_counts()

plt.figure(figsize=(9, 5))
department_count.plot(kind="bar")

plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()

# Save chart
plt.savefig("reports/figures/department_bar_chart.png")

plt.show()

print("Bar chart created and saved successfully!")
```

**Insight:** The bar chart compares employee counts across departments and identifies which departments have the highest and lowest workforce.

```python
# 2. HISTOGRAM - Employee Age Distribution

plt.figure(figsize=(9, 5))

plt.hist(
    df["age"].dropna(),
    bins=10,
    edgecolor="black"
)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.tight_layout()

# Save chart
plt.savefig("reports/figures/age_histogram.png")

plt.show()

print("Histogram created and saved successfully!")
```

**Insight:** Most employees are between 35 and 40 years old, while employee ages are spread across a wider range.

```python
# 3. SCATTER PLOT - Experience vs Salary

plt.figure(figsize=(9, 5))

plt.scatter(
    df["experience_years"],
    df["salary_inr"],
    alpha=0.7
)

plt.title("Employee Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary (INR)")

plt.tight_layout()

# Save chart
plt.savefig("reports/figures/experience_salary_scatter.png")

plt.show()

print("Scatter plot created and saved successfully!")
```

**Insight:** The scatter plot helps identify the relationship between employee experience and salary. It shows how salary varies among employees with different experience levels.

```python
# 4. LINE CHART - Salary Trend by Joining Date

df["joining_date"] = pd.to_datetime(df["joining_date"])

salary_trend = df.sort_values("joining_date")

plt.figure(figsize=(10, 5))

plt.plot(
    salary_trend["joining_date"],
    salary_trend["salary_inr"],
    marker="o",
    label="Employee Salary"
)

plt.title("Salary Trend by Joining Date")
plt.xlabel("Joining Date")
plt.ylabel("Salary (INR)")
plt.xticks(rotation=45)
plt.legend()

plt.tight_layout()

# Save chart
plt.savefig("reports/figures/salary_trend_line.png")

plt.show()

print("Line chart created and saved successfully!")
```

**Insight:** The line chart shows how employee salaries vary across different joining dates and helps identify salary patterns over time.

```python
# 5. BOX PLOT - Salary Distribution

plt.figure(figsize=(9, 5))

plt.boxplot(
    df["salary_inr"].dropna(),
    patch_artist=True
)

plt.title("Employee Salary Distribution")
plt.xlabel("Employees")
plt.ylabel("Salary (INR)")

plt.tight_layout()

# Save chart
plt.savefig("reports/figures/salary_boxplot.png")

plt.show()

print("Box plot created and saved successfully!")
```

**Insight:** The box plot displays the median salary, salary variation and potential outliers among employees.

## Chart Interpretations

### 1. Bar Chart – Employee Count by Department
This chart compares the number of employees across different departments and helps identify departments with higher and lower employee counts.

### 2. Histogram – Employee Age Distribution
This chart shows the distribution of employee ages and helps understand the age range and concentration of employees.

### 3. Scatter Plot – Experience vs Salary
This chart shows the relationship between employee experience and salary. It helps identify whether salary tends to increase with experience.

### 4. Line Chart – Salary Trend by Joining Date
This chart displays employee salaries according to their joining dates and helps observe salary variations over time.

### 5. Box Plot – Salary Distribution
This chart displays the median, spread and potential outliers in employee salaries.
